# **Pegar Planilha**

```python
import pandas as pd
import numpy as np

# Carregar Dados
df = pd.read_excel('caso 2.xlsx', header=1)

# Renomear Colunas
df.columns = [
    'Local',
    'Total_Sem_Fund',
    'Total_Fund_Med',
    'Total_Med_Sup',
    'Total_Superior',
    'Homens_Sem_Fund',
    'Homens_Fund_Med',
    'Homens_Med_Sup',
    'Homens_Superior',
    'Mulheres_Sem_Fund',
    'Mulheres_Fund_Med',
    'Mulheres_Med_Sup',
    'Mulheres_Superior'
]

# Limpeza
df = df.copy()
df['Local'] = df['Local'].astype(str).str.strip()

# Remover Lixo
df = df[df['Local'].notna()]
df = df[df['Local'] != 'nan']
df = df[df['Local'] != '']

# Reset Index
df = df.reset_index(drop=True)

```

# **Dark Mode**

```python
import plotly.graph_objects as go

# DARK MODE
template_dark = go.layout.Template(
    layout=go.Layout(
        paper_bgcolor='#111111',
        plot_bgcolor='#111111',
        font=dict(color='white'),
        title_font=dict(color='white'),

        xaxis=dict(
            color='white',
            gridcolor='#333'
        ),

        yaxis=dict(
            color='white',
            gridcolor='#333'
        ),

        hoverlabel=dict(
            font_color='white',
            font_size=13
        ),

        legend=dict(
            font=dict(color='white')
        )
    )
)
```

# **Gráfico 1 — Brasil (Total / Homens / Mulheres)**

```python
brasil = df[df['Local'] == 'Brasil']
print(df.head())
print(df.columns.tolist())
niveis = [
    'Total_Sem_Fund',
    'Total_Fund_Med',
    'Total_Med_Sup',
    'Total_Superior'
]

labels = ['Sem Instr.', 'Fund/Médio', 'Médio/Sup', 'Superior']

fig = go.Figure()

fig.add_trace(go.Bar(
    name='Total',
    x=labels,
    y=brasil[niveis].values.flatten(),
    marker_color = '#ff7f0e' # laranja
))

fig.add_trace(go.Bar(
    name='Homens',
    x=labels,
    y=brasil[[c.replace('Total', 'Homens') for c in niveis]].values.flatten(),
    marker_color = '#1f77b4'  # azul
))

fig.add_trace(go.Bar(
    name='Mulheres',
    x=labels,
    y=brasil[[c.replace('Total', 'Mulheres') for c in niveis]].values.flatten(),
    marker_color='#e83e8c'  # rosa
))

fig.update_layout(
    template=template_dark,
    barmode='group',
    title='Brasil - Escolaridade por Gênero'
)

fig.show()
```

Observa-se que a maior concentração da população brasileira está nos níveis intermediários de escolaridade, com destaque para o ensino médio completo. No entanto, ainda há uma parcela significativa com baixa escolaridade (sem instrução ou fundamental incompleto), o que evidencia desafios estruturais na educação.

Ao comparar homens e mulheres, percebe-se que as mulheres apresentam maior percentual de ensino superior completo, enquanto os homens concentram maiores índices nos níveis mais baixos de escolaridade.

# **Gráfico 2 — Comparação das Regiões**

```python
regioes = ['Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']
df_reg = df[df['Local'].isin(regioes)].set_index('Local')

fig = go.Figure()

for nivel, label in zip(niveis, labels):
    fig.add_trace(go.Bar(
        name=label,
        x=df_reg.index,
        y=df_reg[nivel]
    ))

fig.update_layout(
    template=template_dark,
    barmode='group',
    title='Regiões x Escolaridade'
)

fig.show()
```

A análise regional revela diferenças significativas na distribuição da escolaridade no Brasil. As regiões Sudeste e Sul apresentam maiores percentuais de ensino superior completo, indicando maior desenvolvimento educacional.

Por outro lado, as regiões Norte e Nordeste concentram maiores índices de baixa escolaridade, evidenciando desigualdades regionais históricas no acesso à educação.

# **Gráfico 3 — Homens vs Mulheres nas regiões**

```python
fig = go.Figure()

fig.add_trace(go.Bar(
    name='Homens',
    x=df_reg.index,
    y=df_reg['Homens_Superior'],
    marker_color='#ff7f0e'
))

fig.add_trace(go.Bar(
    name='Mulheres',
    x=df_reg.index,
    y=df_reg['Mulheres_Superior'],
    marker_color='#e83e8c'
))

fig.update_layout(
    template = template_dark,
    barmode = 'group',
    title = 'Superior Completo por Gênero'
)

fig.show()
```

Ao comparar homens e mulheres nas regiões, observa-se um padrão consistente: as mulheres apresentam maior percentual de ensino superior completo em praticamente todas as regiões.

Esse comportamento indica uma tendência de maior permanência feminina no sistema educacional, enquanto os homens apresentam maior concentração nos níveis mais baixos de escolaridade.

# **Estatística Descritiva**

```python
macro = ['Brasil', 'Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']
estados = df[~df['Local'].isin(macro)]

dados = estados['Total_Superior']

print('Média:', dados.mean())
print('Mediana:', dados.median())
print('Desvio:', dados.std())
print('Min:', dados.min())
print('Max:', dados.max())
print('Amplitude:', dados.max() - dados.min())
```

As medidas estatísticas evidenciam a existência de variação significativa entre os estados brasileiros em relação ao ensino superior completo. O desvio padrão indica dispersão relevante dos dados, sugerindo desigualdade educacional entre as unidades da federação.

Estados com valores mais elevados destacam-se positivamente, enquanto aqueles com menores percentuais indicam regiões mais vulneráveis, reforçando a heterogeneidade do país.

# **Gráfico 4 — Ranking dos Estados**

```python
estados_ord = estados.sort_values('Total_Superior')

fig = go.Figure()

fig.add_trace(go.Bar(
    x=estados_ord['Local'],
    y=estados_ord['Total_Superior']
))

fig.update_layout(
    template=template_dark,
    title='Ranking dos Estados (Superior Completo)'
)

fig.show()
```

O ranking dos estados evidencia quais unidades federativas apresentam melhores e piores indicadores de escolaridade superior. Observa-se uma concentração dos melhores resultados em regiões mais desenvolvidas, enquanto estados com menor percentual de ensino superior refletem desafios estruturais.

Essa desigualdade reforça a necessidade de políticas públicas direcionadas para regiões mais vulneráveis.

# **Nordeste Superior Completo**

```python
nordeste_estados = [
    'Bahia','Pernambuco','Ceará','Maranhão',
    'Paraíba','Piauí','Alagoas','Sergipe','Rio Grande do Norte'
]

df_ne = estados[estados['Local'].isin(nordeste_estados)]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=df_ne['Local'],
    y=df_ne['Total_Superior']
))

fig.update_layout(
    template=template_dark,
    title='Nordeste - Superior Completo'
)

fig.show()
```

A análise específica da região Nordeste evidencia níveis mais baixos de ensino superior em comparação com outras regiões do país. Isso confirma o padrão observado anteriormente de maior vulnerabilidade educacional na região.

Mesmo assim, há variações internas entre os estados, indicando que a desigualdade não ocorre apenas entre regiões, mas também dentro delas.

# **MUNICÍPIOS (BAHIA)**

```python
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

# Bahia
municipios = df[df['Local'].str.contains('Bahia', case=False, na=False)]

dados_mun = municipios['Total_Superior']


print('=== Estatísticas - Ensino Superior (Bahia) ===')
print('Média:', dados_mun.mean())
print('Mediana:', dados_mun.median())
print('Desvio padrão:', dados_mun.std())
print('Mínimo:', dados_mun.min())
print('Máximo:', dados_mun.max())
print('Amplitude:', dados_mun.max() - dados_mun.min())


# Histograma
fig = px.histogram(
    dados_mun,
    nbins=10,
    title='Distribuição - Ensino Superior (Municípios da Bahia)',
    template='plotly_dark',
    labels={
        'value': 'Quantidade',
        'count': 'Frequência'
    }
)

fig.show()
```

A análise dos municípios revela a existência de desigualdade interna dentro do estado analisado. A diferença entre valores mínimos e máximos, juntamente com o desvio padrão, indica que nem todos os municípios apresentam o mesmo nível de escolaridade.

Isso demonstra que, mesmo dentro de um único estado, há disparidades significativas no acesso à educação.

# **Histograma**

```python
fig = go.Figure()

fig.add_trace(go.Histogram(x=dados_mun))

fig.update_layout(
    template=template_dark,
    title='Distribuição dos Municípios (Desigualdade)'
)

fig.show()
```

O histograma mostra a distribuição dos níveis de escolaridade entre os municípios. Observa-se a concentração dos dados em determinadas faixas, indicando que a maioria dos municípios apresenta níveis semelhantes de escolaridade.

No entanto, a presença de valores mais extremos reforça a existência de desigualdade educacional dentro do estado.

# **Radar**

```python
fig = go.Figure()

for regiao in df_reg.index:
    fig.add_trace(go.Scatterpolar(
        r=df_reg.loc[regiao, niveis].values,
        theta=labels,
        fill='toself',
        name=regiao
    ))

fig.update_layout(
    template=template_dark,
    polar=dict(bgcolor='#111111'),
    title='Radar das Regiões'
)

fig.show()
```

O gráfico radar permite visualizar de forma comparativa o perfil educacional das regiões. Nota-se que algumas regiões apresentam maior equilíbrio entre os níveis de escolaridade, enquanto outras concentram maior percentual em níveis mais baixos.

Essa visualização facilita a identificação de padrões regionais e reforça as desigualdades existentes no país.

# **PERFIL POR REGIÃO**

```python
fig = go.Figure()

for regiao in df_reg.index:
    fig.add_trace(go.Bar(
        name=regiao,
        x=labels,
        y=df_reg.loc[regiao, niveis]
    ))

fig.update_layout(
    template=template_dark,
    barmode='group',
    title='Perfil Educacional por Região'
)

fig.show()
```

# **RISCO EDUCACIONAL POR REGIÃO**

```python
import pandas as pd


regioes = ['Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']

df_reg = (
    df[df['Local'].isin(regioes)]
    .copy()
    .set_index('Local')
)


df_reg['Risco'] = df_reg['Total_Sem_Fund']


fig = go.Figure()

fig.add_trace(go.Bar(
    x=df_reg.index,
    y=df_reg['Risco'],
    marker_color='#ff7f0e'
))

fig.update_layout(
    template=template_dark,
    title='Risco Educacional por Região (Baixa Escolaridade)',
    xaxis_title='Região',
    yaxis_title='População em Risco',
    hoverlabel=dict(
        bgcolor='orange',
        font_color='white'
    )
)

fig.show()
```

O gap educacional evidencia a diferença entre a população com ensino superior completo e aquela com baixa escolaridade. Observa-se que todas as regiões apresentam valores negativos, indicando predominância de baixa escolaridade.

Esse resultado revela um cenário estrutural de desigualdade educacional no Brasil, impactando diretamente o desenvolvimento socioeconômico e a qualificação da população para o mercado de trabalho.
